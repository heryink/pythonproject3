```python

students = pd.DataFrame({
    "Name": [
        "Ada", "Tunde", "Mariam", "David", "Ayomide",
        "Emeka", "Sarah", "Daniel", "Aisha", "Michael"
    ],
    "Age": [
        20, 21, np.nan, 22, 18,
        23, 21, np.nan, 22, 24
    ],
    "Department": [
        "Computer Science", "Mathematics", "Computer Science", np.nan,
        "Software Engineering", "biology", "Mathematics", "Computer Science",
        np.nan, "Physics"
    ],
    "GPA": [
        4.5, np.nan, 4.2, 3.8, 4.7,
        np.nan, 4.1, 3.9, 4.6, 4.0
    ],
    "Level": [
        200, 300, 200, 400, 300,
        300, 200, 400, 300, np.nan
    ],
    "Scholarship": [
        "Yes", "No", np.nan, "No", "Yes",
        "No", "Yes", np.nan, "Yes", "No"
    ]
})

students
```

```python
#Q2
print(students.isnull())
print(f"total number of missing element: {students.isnull().sum().sum()}")
print(students.copy().dropna())
data = students.copy()
data["GPA"] = data["GPA"].fillna(data["GPA"].mean())
data["Level"] = data["Level"].fillna(data["Level"].median())
data["Scholarship"] = data["Scholarship"].replace(np.nan,"pending")
print(data)
```

```python
#Q3
Age_filter = data[data["Age"]< 23]
print(Age_filter)
print(f"\n\n")
scholarship_f = data[data["Scholarship"]=="pending"]
print(scholarship_f)
print(f"\n\n")
data_filtered = data[(data["GPA"]>4.49) & (data["Level"]>=100)]
print(f"\n\n")
print(data_filtered)
```

```python
#Q4
print(f"mean age: {data["Age"].mean()}")
print(f"median GPA: {data["GPA"].median()}")
print(f"standard deviation of GPA: {data['GPA'].std():.2f}")
print(data["Scholarship"].value_counts())
print(f"correlation between Age and level{data["Age"].corr(data["Level"]):.4f}")
```

```python
#Q5
sorted_data = data.sort_values(by="Age",ascending=False)
print(sorted_data)
print(f"\n\n")
print(data.groupby("Scholarship")["GPA"].mean())  #groupby create another set of dataframes internally based on the categories of scholarship then it looks for the mean GPA of dataframes/categories
```

```python
#Q1
import pandas as pd
import numpy as np

titanic_dataset = pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv")
print(titanic_dataset.shape)
print(titanic_dataset.columns)
print(titanic_dataset.dtypes)
print(titanic_dataset.isnull().sum())            # the deck columns has the highest missing values
print(titanic_dataset.head())
```

```python
#Q2 & #Q3
print(titanic_dataset.describe())  #mean and max of the fare column: 32.204208, 512.329200 respectively
#considering the fact that the median is less than the mean iin a relatively large amount, the fare column values is skewed to the right
print(f"Survival rate: {titanic_dataset["survived"].mean():.2f}")
print(f"Passengers in each classes: {titanic_dataset['pclass'].value_counts()}")
print(f"Average Age of passengers: {titanic_dataset['age'].mean():.1f}")
```

```python
#Q4
print(f"Survival rate by sex: {titanic_dataset.groupby("sex")["survived"].mean()}")
print(f"Survival rate by passenger class: {titanic_dataset.groupby("pclass")["survived"].mean()}")
#according to the results females and passengers in class 1 are more likely to survive
```

```python
#Q5
print(f"Correlation between fare and survived: {titanic_dataset["fare"].corr(titanic_dataset["survived"]):.4f}")
print(f"Correlation between fare and passenger class: {titanic_dataset["pclass"].corr(titanic_dataset["fare"]):.4f}")
#I would expect pclass and fare to be negatively correlated because higher-class
# passengers pay higher fares. which is inversely proportional
```

```python
#Q1
netflix_dataset = pd.read_csv("https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/netflix_titles.csv")
print(netflix_dataset.shape)
netflix_dataset.info()
print(f"Sum of missing element in each column: {netflix_dataset.isnull().sum()}")
```

```python
#Q2
netflix_dataset_dup = netflix_dataset.copy()
netflix_dataset_dup.drop(columns="director", inplace=True)

netflix_dataset_dup["rating"] = netflix_dataset_dup["rating"].fillna("Not Rated")
netflix_dataset_dup["cast"] = netflix_dataset_dup["cast"].fillna("Unknown")
netflix_dataset_dup["country"] = netflix_dataset_dup["country"].fillna("Unknown")

netflix_dataset_dup.dropna(subset=["date_added"], inplace=True)

print(netflix_dataset_dup.isnull().sum())
```

```python
#Q3
print(netflix_dataset_dup.head(5))
print(netflix_dataset_dup["duration"].describe())
print(netflix_dataset_dup["release_year"].describe())
print(netflix_dataset_dup["date_added"].describe())
```

```python
#Q4
#statistics on duration column
movies = netflix_dataset_dup[netflix_dataset_dup["type"] == "Movie"]["duration"]
movies_mins = movies.str.replace(" min", "").astype(int)
print(f"Average movie duration: {movies_mins.mean():.2f} mins")

#year with highest release:
print(f"Year with highest release: {netflix_dataset_dup['release_year'].value_counts().index[0]}")

#number of release based on show type
print(netflix_dataset_dup["rating"].value_counts())


print(f"Cleaned dataset shape: {netflix_dataset_dup.shape}")




# The average movie duration of 99 mins since most movies usually between 1hr 2o min and 1hr 4o min

# 2018 being the year with the most releases makes sense since it was this moment were aggressive in their campaigns and trying to reach as much people as possible, this year also marked their first milestone in revenue

# TV-MA and TV-14 is the most common rating (2025 titles), which is expected since Netflix primarily targets adult and youth audiences.
```

```python
#Q5

netflix_dataset_dup.to_csv("Netflix_dataset", index=False)

#what i did to the dataset:
#I dropped The 'director' column  because it had way too many missing values more than 20%
# For 'cast' and 'country', some titles genuinely don't have that info  available, so I filled those with 'Unknown' instead of losing the rows.
# 'rating' only had 9 missing values, so I filled those with 'Not Rated'
# 'date_added' had 11 missing rows — small enough to just drop without
# losing anything significant.
# For filtering, I checked that every row had a valid type — either 'Movie'
# or 'TV Show'. Nothing got removed, which just means the data was already
# clean on that front.
```
